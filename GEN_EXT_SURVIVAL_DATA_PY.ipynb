```python
m1 = {
    '2016_TCGA':'2016_TCGA_rna_count_ssgsea.dfx',
    '2018_Cancer_cell':'2018_Cancer_cell_rna_rpkm_ssgsea.dfx',
    '2019_PNAS':'2019_PNAS_rna_tpm_ssgsea.dfx',
    '2023_Mol_Cell':'2023_Mol_Cell_rna_tpm_ssgsea.dfx',
    '2024_CRM':'2024_CRM_prt_ssgsea.dfx',
    '2024_natcan_prt':'2024_natcan_prt_ssgsea.dfx',
    '2024_natcan_rna':'2024_natcan_rna_tpm_ssgsea.dfx',
    '2019_CancerCell_prt':"2019_CancerCell__prt.dfx",
    '2019_CancerCell_rna':"2019_CancerCell__rna_normd.dfx",
}
m2 = {
    '2016_TCGA':'2016_TCGA__clin.dfx',
    '2018_Cancer_cell':'2018_Cancer_cell__clin.dfx',
    '2019_PNAS':'2019_PNAS__clin.dfx',
    '2023_Mol_Cell':'2023_Mol_Cell__clin.dfx',
    '2024_CRM':'2024_CRM__clin.dfx',
    '2024_natcan_prt':'2024_natcan__prt_clin.dfx',
    '2024_natcan_rna':'2024_natcan__rna_clin.dfx',
    '2019_CancerCell_prt':"2019_CancerCell__prt_clin.dfx",
    '2019_CancerCell_rna':"2019_CancerCell__rna_clin.dfx",
}
```

```python
import joblib
OBJ = joblib.load(filename='export_c3/fig4_gsea_rf_model_evenly_dis')
print(OBJ.keys())
dtc = OBJ['dtc']
matv_ss = OBJ['matv_ss']
```

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
def standarize(mat):
    """
    mat: index are ssGSEA terms, columns are patients
    Returns:
        StandardScaler scaled dataframe, index are Patients, columns are ssGSEA terms
    """
    mt = mat.T
    x = StandardScaler().fit_transform(mt)
    res = pd.DataFrame(x, index = mt.index, columns=mt.columns)
    return res
import joblib
CACHE = J('export_c3/ext_data_valid_pca_surv_data.joblib')
if E(CACHE):
    print(f'=====\n Load existing cache {CACHE}')
    res = joblib.load(CACHE)
else:
    res = {}
    for name in m1.keys():
        file = m1[name]
        mat = db.importing(J(base, file))
        fclin = m2[name]
        clin = db.importing(J(base, fclin))
        mask = clin.columns.str.endswith('MONTHS')
        s = pd.Series(clin.columns)
        s[mask] = s[mask].str.replace('MONTHS','MONTH', regex=False)
        clin.columns = s
        mask = s.str.endswith('_MONTH')
        os_types = s[mask].str.removesuffix('_MONTH').to_list()
        mask_cols = (s=='id')|(s.str.endswith('_MONTH')|(s.str.endswith('_STATUS')))


        pca = PCA(n_components=4)
        X_r = pd.DataFrame(pca.fit_transform(mat.T),index=mat.columns ,columns=['X1','X2','X3','X4'])

        mat_ss = standarize(mat)
        cls = dtc.predict(mat_ss.loc[:,dtc.feature_names_in_])

        df = pd.DataFrame({"id":mat.columns,"Cluster":cls})

        data_pca = pd.merge( X_r[['X1','X2']]
                            , df,
                    left_index=True,right_on='id',how='left')
        data_pca = pd.merge(data_pca,clin.loc[:,mask_cols.values],on='id',how='left')

        res[name] = {"data_pca":data_pca,'os_types':os_types}

        print(file, data_pca.groupby('Cluster')['id'].count())
    print(f'=====\n Dump cache to {CACHE}')
    joblib.dump(res, CACHE)
```
