```r
get_df = function (hm, name = c("C1",'C3','C2')){
    col_ord  = column_order(hm)   # list of length 3, each = integer col indices

    # indices are relative to the matrix passed to Heatmap (== colnames(mat.sel))
    col_groups = lapply(col_ord, function(idx) colnames(r$mat)[idx])
    print(col_groups)
    names(col_groups) = name
    
    df <- data.frame(
      tsb = unlist(col_groups, use.names = FALSE),
      group  = rep(names(col_groups), lengths(col_groups))
    )
    df
}
```

```r
eval_pred = function(truth, pred, classes = c("C1", "C2", "C3"),
                      weighted_kappa = TRUE,
                      kappa_weights  = c("quadratic", "linear")) {
  kappa_weights = match.arg(kappa_weights)

  truth = factor(truth)
  pred  = factor(pred, levels = levels(truth))
  lev   = levels(truth)
  K     = length(lev)

  cm = table(GroundTruth = truth, Prediction = pred)   # 行=真值, 列=预测
  N  = sum(cm)

  rs = rowSums(cm); cs = colSums(cm)
  TP = diag(cm); FP = cs - TP; FN = rs - TP; TN = N - TP - FP - FN

  # ---- 每类 (one-vs-rest) ----
  precision   = TP / (TP + FP)
  recall      = TP / (TP + FN)
  specificity = TN / (TN + FP)
  npv         = TN / (TN + FN)
  f1          = 2 * precision * recall / (precision + recall)
  bal_acc     = (recall + specificity) / 2

  # ---- 整体 ----
  accuracy = sum(TP) / N
  pe    = sum((rs / N) * (cs / N))
  kappa = (accuracy - pe) / (1 - pe)

  kappa_w = NA_real_
  if (weighted_kappa && K > 1) {
    idx = seq_len(K)
    D = outer(idx, idx, function(i, j)
      if (kappa_weights == "quadratic") (i - j)^2 / (K - 1)^2
      else abs(i - j) / (K - 1))
    kappa_w = 1 - sum(D * (cm / N)) / sum(D * outer(rs / N, cs / N))
  }
  mcc = (N * sum(TP) - sum(cs * rs)) /
         sqrt((N^2 - sum(cs^2)) * (N^2 - sum(rs^2)))

  sel = intersect(classes, lev)

  # ===== data.frame 1: 混淆矩阵 =====
  cm_df = as.data.frame.matrix(cm)

  # ===== data.frame 2: metrics, 每个指标一行 =====
  cls_mat = rbind(PPV = precision, Recall = recall, Specificity = specificity,
                   NPV = npv, F1 = f1, BalancedAcc = bal_acc)[, lev, drop = FALSE]
  macro = rowMeans(cls_mat[, sel, drop = FALSE], na.rm = TRUE)

  per_class_df = data.frame(Metric = rownames(cls_mat), cls_mat,
                             Summary = macro, row.names = NULL, check.names = FALSE)

  overall_df = data.frame(Metric = c("Accuracy", "Kappa", "KappaW", "MCC"),
                           Summary = c(accuracy, kappa, kappa_w, mcc))
  for (cl in lev) overall_df[[cl]] = NA_real_
  overall_df = overall_df[, c("Metric", lev, "Summary")]

  metrics = rbind(per_class_df, overall_df)
  metrics[, -1] = round(metrics[, -1], 4)
  rownames(metrics) = NULL

  list(cm = cm_df, metrics = metrics)
}
```
