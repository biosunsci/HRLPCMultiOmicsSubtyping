### Fig2 pre ranked GSEA

```python
import gseapy
```

```python
ar = pd.read_excel('external_data/AR调控轴基因总结.xlsx', sheet_name = 'Summarize')
ar
```

```python
gene_sets = {}
for s in ar.columns:
    genes = ar[s].dropna().to_list()
    gene_sets[s] = genes
```

```python
gene_sets.keys()
```

#### prt

```python
dap = pd.read_excel('report_c3/gsea_input/fig.2C_data_DEG_DAP.xlsx',sheet_name = 'Protein_C1')
dap
```

```python
ranked_genes_prt = dap.set_index("symbol")['B'].sort_values(ascending=False)
ranked_genes_prt
```

```python
## this is additional check
## not used in figures
gseapy.prerank(
    rnk=ranked_genes_prt,        # Pre-ranked gene list
    gene_sets=gene_sets,     # Your custom gene set
    outdir='./report_c3/gsea_report/prt_c1/',
    permutation_num=1000,    # Number of permutations for p-value
    min_size=5,              # Minimum genes in a gene set
    max_size=500,            # Maximum genes in a gene set
    format='pdf',           # Output figure format
    seed=42                 # Random seed for reproducibility
)
```

#### rna

```python
deg = pd.read_excel('report_c3/gsea_input/fig.2C_data_DEG_DAP.xlsx',sheet_name = 'RNAseq_C1')
deg['stat'] = deg['stat'].fillna(0)
deg
```

```python
ranked_genes_rna = deg.set_index("symbol")['stat'].sort_values(ascending=False)
ranked_genes_rna
```

```python
## this is Fig.2C used
gseapy.prerank(
    rnk=ranked_genes_rna,        # Pre-ranked gene list
    gene_sets=gene_sets,     # Your custom gene set
    outdir='./report_c3/gsea_report/rna_c1/',
    permutation_num=1000,    # Number of permutations for p-value
    min_size=5,              # Minimum genes in a gene set
    max_size=500,            # Maximum genes in a gene set
    format='pdf',           # Output figure format
    seed=42                 # Random seed for reproducibility
)
```
